**B1. Retorno multiple**

Escriba estadisticas(datos) que devuelva el minimo, el maximo y el promedio de una lista en una sola pasada.

```python
def estadisticas(datos):
    minimo = maximo = datos[0]
    total = 0
    for x in datos:
        if x < minimo:
            minimo = x
        if x > maximo:
            maximo = x
        total += x
    return minimo, maximo, total / len(datos)

print(estadisticas([4, 7, 1, 9, 3]))
```

**B2. Parametro por defecto**

Escriba saludar(nombre, saludo="Hola") que retorne el saludo compuesto; pruebe con y sin el segundo argumento.

**B3. Factorial recursivo**

Implemente factorial(n) de forma recursiva (con caso base) y pruebelo con n=5.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

**B4. Suma de digitos recursiva**

Implemente suma_digitos(n) recursiva: suma_digitos(347) -> 14.

```python
def suma_digitos(n):
    if n < 10:
        return n
    return n % 10 + suma_digitos(n // 10)

print(suma_digitos(347))
```

**B5. Frecuencia de palabras**

Dada una frase, construya un diccionario {palabra: veces} normalizando a minusculas.

```python
frase = "el perro y el gato y el raton"
frecuencia = {}
for palabra in frase.lower().split():
    frecuencia[palabra] = frecuencia.get(palabra, 0) + 1
print(frecuencia)
```

**B6. Operaciones de conjuntos**

Dadas dos listas de estudiantes, use conjuntos para hallar quienes estan en ambas, en total (sin repetir) y solo en la primera.

```python
a = ["Ana","Luis","Marta"]
b = ["Luis","Pedro","Ana"]
print("En ambas:", set(a) & set(b))
print("En total:", set(a) | set(b))
print("Solo en la primera:", set(a) - set(b))
```
