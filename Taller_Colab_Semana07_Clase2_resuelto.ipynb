**B1. Promedio por fila**

Escriba una funcion promedio_filas(M) que devuelva una lista con el promedio de cada fila de la matriz.

```python
def promedio_filas(M):
    resultado = []
    for i in range(len(M)):
        suma = 0
        for j in range(len(M[i])):
            suma = suma + M[i][j]
        promedio = suma / len(M[i])
        resultado.append(promedio)
    return resultado

print(promedio_filas([[4,3,2],[1,5,9]]))
```

**B2. Maximo y su posicion**

Encuentre el valor maximo de una matriz y la posicion (fila, columna) donde se encuentra.

```python
M = [[4,3,2],[1,9,5],[7,8,6]]
maximo = M[0][0]
fila_max = 0
col_max = 0
for i in range(len(M)):
    for j in range(len(M[i])):
        if M[i][j] > maximo:
            maximo = M[i][j]
            fila_max = i
            col_max = j
print("Maximo:", maximo)
print("Posicion: fila", fila_max, "columna", col_max)
```

**B3. Crear e imprimir matriz**

Lea f y c, cree una matriz f x c leida por teclado y muestrela con formato tabular (filas y columnas alineadas).

```python
f = int(input("Filas: "))
c = int(input("Columnas: "))
M = []
for i in range(f):
    fila = []
    for j in range(c):
        valor = int(input("Elemento " + str(i) + "," + str(j) + ": "))
        fila.append(valor)
    M.append(fila)
for i in range(f):
    for j in range(c):
        print(M[i][j], end="\t")
    print()
```

**B4. Sumas por fila y columna**

Dada una matriz, calcule e imprima la suma de cada fila y la suma de cada columna.

```python
M = [[4,3,2],[1,5,9],[7,8,6]]
filas = len(M)
columnas = len(M[0])
for i in range(filas):
    suma = 0
    for j in range(columnas):
        suma = suma + M[i][j]
    print("Suma fila", i, "=", suma)
for j in range(columnas):
    suma = 0
    for i in range(filas):
        suma = suma + M[i][j]
    print("Suma columna", j, "=", suma)
```

**B5. Cuadrado magico**

Verifique si una matriz 3x3 es un cuadrado magico (todas las filas, columnas y diagonales suman lo mismo).

```python
M = [[2,7,6],[9,5,1],[4,3,8]]
n = 3
objetivo = M[0][0] + M[0][1] + M[0][2]
es_magico = True
for i in range(n):
    suma_fila = 0
    suma_col = 0
    for j in range(n):
        suma_fila = suma_fila + M[i][j]
        suma_col = suma_col + M[j][i]
    if suma_fila != objetivo or suma_col != objetivo:
        es_magico = False
diag1 = 0
diag2 = 0
for i in range(n):
    diag1 = diag1 + M[i][i]
    diag2 = diag2 + M[i][n - 1 - i]
if diag1 != objetivo or diag2 != objetivo:
    es_magico = False
if es_magico:
    print("Es cuadrado magico")
else:
    print("No es cuadrado magico")
```

**B6. Transponer**

Escriba una funcion transponer(M) que devuelva la matriz transpuesta (T[j][i] = M[i][j]).

```python
def transponer(M):
    filas = len(M)
    columnas = len(M[0])
    T = []
    for j in range(columnas):
        nueva_fila = []
        for i in range(filas):
            nueva_fila.append(M[i][j])
        T.append(nueva_fila)
    return T

print(transponer([[1,2,3],[4,5,6]]))
```
