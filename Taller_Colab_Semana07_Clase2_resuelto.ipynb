**B1. Promedio por fila**

Escriba una funcion promedio_filas(M) que devuelva una lista con el promedio de cada fila de la matriz.

```python
def promedio_filas(M):
    return [sum(fila) / len(fila) for fila in M]

print(promedio_filas([[4,3,2],[1,5,9]]))
```

**B2. Maximo y su posicion**

Encuentre el valor maximo de una matriz y la posicion (fila, columna) donde se encuentra.

```python
M = [[4,3,2],[1,9,5],[7,8,6]]
maximo = M[0][0]
pos = (0, 0)
for i in range(len(M)):
    for j in range(len(M[i])):
        if M[i][j] > maximo:
            maximo = M[i][j]
            pos = (i, j)
print("Maximo:", maximo, "en la posicion", pos)
```

**B3. Crear e imprimir matriz**

Lea f y c, cree una matriz f x c leida por teclado y muestrela con formato tabular (filas y columnas alineadas).

```python
f = int(input("Filas: "))
c = int(input("Columnas: "))
M = []
for i in range(f):
    fila = []
    for j in range(c):
        fila.append(int(input(f"Elemento [{i}][{j}]: ")))
    M.append(fila)
for fila in M:
    for x in fila:
        print(f"{x:5}", end="")
    print()
```

**B4. Sumas por fila y columna**

Dada una matriz, calcule e imprima la suma de cada fila y la suma de cada columna.

```python
M = [[4,3,2],[1,5,9],[7,8,6]]
for i in range(len(M)):
    print("Suma fila", i, ":", sum(M[i]))
for j in range(len(M[0])):
    suma = 0
    for i in range(len(M)):
        suma += M[i][j]
    print("Suma columna", j, ":", suma)
```

**B5. Cuadrado magico**

Verifique si una matriz 3x3 es un cuadrado magico (todas las filas, columnas y diagonales suman lo mismo).

```python
M = [[2,7,6],[9,5,1],[4,3,8]]
n = len(M)
objetivo = sum(M[0])
sumas = [sum(fila) for fila in M]
sumas += [sum(M[i][j] for i in range(n)) for j in range(n)]
sumas.append(sum(M[i][i] for i in range(n)))
sumas.append(sum(M[i][n-1-i] for i in range(n)))
es_magico = all(s == objetivo for s in sumas)
print("Es cuadrado magico" if es_magico else "No es cuadrado magico")
```

**B6. Transponer**

Escriba una funcion transponer(M) que devuelva la matriz transpuesta (T[j][i] = M[i][j]).

```python
def transponer(M):
    T = []
    for j in range(len(M[0])):
        fila = []
        for i in range(len(M)):
            fila.append(M[i][j])
        T.append(fila)
    return T

print(transponer([[1,2,3],[4,5,6]]))
```
